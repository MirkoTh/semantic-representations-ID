### extract dimensional weights per ndim and calculate reliabilities

```python
def shrel_reorder_args(idxs, ndims, l_splithalf):
    df_sh, df_corr, g = ut.split_half_reliabilities(l_splithalf, idxs, ndims, do_plot=False)
    return df_sh, df_corr, g
sh_partial = partial(shrel_reorder_args, l_splithalf=l_splithalf)
l_dicts = [dict_idxs_list_use[idx] for idx in range_dims]
l_split_half_reliabilities = list(map(sh_partial, l_dicts, range_dims))
```

```python
l_corrs = [l[1] for l in l_split_half_reliabilities]
df_corrs = reduce(lambda x, y: pd.concat([x, y]), l_corrs)

# spearman-brown prophecy correction
df_corrs["r_corrected"] = (2*df_corrs["r"])/(1 + df_corrs["r"])

f, ax = plt.subplots(1)

ax.fill_between([8.5, 12], 0, 1, color='red', alpha=0.3, label="No unique mapping")

# Violin plot
sns.violinplot(data=df_corrs, x="ndims", y="r_corrected", hue="ndims", inner=None, legend=False, ax=ax)

# Swarm plot on top
sns.swarmplot(data=df_corrs, x="ndims", y="r_corrected", color="white", legend=False, ax=ax, size=3.5)

ax.axhline(y=.7, color='lightgreen', linestyle='--', linewidth=2, label='THX Group-Level Comparison')
ax.axhline(y=.8, color='darkgreen', linestyle='--', linewidth=2, label='THX Individual-Level Comparison')
ax.axvline(x=8.5, color="red", linestyle='-.', linewidth=2.5, label="")

ax.set_xlabel("Nr. Embedding Dimensions")
ax.set_ylabel("Spearman-Brown Correction")


ax.set_title(ax.get_title(), fontsize=16)
ax.set_xlabel(ax.get_xlabel(), fontsize=14)
ax.set_ylabel(ax.get_ylabel(), fontsize=14)
ax.tick_params(axis='both', labelsize=12)


# Remove duplicate legend
ax.legend()
ax.set_ylim((0, 1))
_ = ax.set_title("Split-Half Reliability by Number of Dimensions")

ax.grid(True)


f.set_size_inches(8, 5)
```

### extract dimensional weights per ndim and calculate reliabilities

```python
sh_partial = partial(shrel_reorder_args, l_splithalf=l_splithalf_v2)
l_dicts_v2 = [dict_idxs_list_use_v2[idx] for idx in range_dims]
l_split_half_reliabilities_v2 = list(map(sh_partial, l_dicts_v2, range_dims))
```

```python
# calculate the split-half reliability for the gini coefficient and the average weight
dict_rels = dict()
for idx, d in enumerate(range_dims):

    df_sh_use = l_split_half_reliabilities[idx][0]
    df_sh_use_v2 = l_split_half_reliabilities_v2[idx][0]

    
    # 2 values < 0 (i.e. very close to zero around -1e-5)
    df_sh_use.loc[df_sh_use["decision_weight_1"] < 0, "decision_weight_1"] = 0
    df_sh_use.loc[df_sh_use["decision_weight_2"] < 0, "decision_weight_2"] = 0
    
    # gini
    ginis = df_sh_use.groupby("id")[["decision_weight_1", "decision_weight_2"]].agg(ut.gini).reset_index()
    ginis.columns = ["id", "gini_half_1", "gini_half_2"]
    r_gini = np.corrcoef(ginis["gini_half_1"], ginis["gini_half_2"])[0, 1]
    r_gini_corrected = (2*r_gini)/(1 + r_gini)
    
    # avg weight
    avg_weight = df_sh_use.groupby("id")[["decision_weight_1", "decision_weight_2"]].agg("mean").reset_index()
    r_avg_weight = np.corrcoef(avg_weight["decision_weight_1"], avg_weight["decision_weight_2"])[0, 1]
    r_avg_weight_corrected = (2*r_avg_weight)/(1 + r_avg_weight)
    
    # avg abs delta
    df_sh_use["delta_decision_weight"] = df_sh_use["decision_weight_1"] - df_sh_use["decision_weight_2"]
    df_sh_use["abs_delta_decision_weight"] = np.abs(df_sh_use["delta_decision_weight"])
    avg_abs_delta = df_sh_use.groupby("id")["abs_delta_decision_weight"].agg("mean").reset_index()
    df_sh_use_v2["delta_decision_weight"] = df_sh_use_v2["decision_weight_1"] - df_sh_use_v2["decision_weight_2"]
    df_sh_use_v2["abs_delta_decision_weight"] = np.abs(df_sh_use_v2["delta_decision_weight"])
    avg_abs_delta_v2 = df_sh_use_v2.groupby("id")["abs_delta_decision_weight"].agg("mean").reset_index()
    avg_abs_delta_both = avg_abs_delta.merge(avg_abs_delta_v2, on="id", how="left", suffixes = ["_v1", "_v2"])
    avg_abs_delta_corr = np.corrcoef(avg_abs_delta_both["abs_delta_decision_weight_v1"], avg_abs_delta_both["abs_delta_decision_weight_v2"])[0, 1]
    # correction formula
    r_avg_delta = (2*avg_abs_delta_corr)/(1 + avg_abs_delta_corr)

    d_result = {"sh_gini":r_gini_corrected, "sh_avg_weight":r_avg_weight_corrected, "avg_abs_delta":r_avg_delta}
    dict_rels[d] = d_result

df_splithalf_summary = pd.DataFrame.from_dict(dict_rels, orient='index').reset_index()
df_splithalf_summary.columns = ["ndim", "gini", "avg_weight", "avg_abs_delta"]
```
