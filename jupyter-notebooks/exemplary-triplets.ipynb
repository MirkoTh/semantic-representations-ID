```python
import os
import re

from functools import partial

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

import matplotlib.colors as mcolors
```

```python
l_concepts = os.listdir("data/images/")
l_m = [re.match("^[a-z]", l_concepts[i]) for i in range(0, len(l_concepts))]
l_filter = [l_m[i] != None for i in range(0, len(l_concepts))]
l_concepts_filtered = [value for value, flag in zip(l_concepts, l_filter) if flag]
f_partial = partial(ut.extract_image, l_concepts_filtered)
```

```python
indices = np.where(np.isin(np.array(l_concepts_filtered), ["almond", "hammock", "peanut_butter", "airplane", "beanie", "carrot", "chips", "cupcake", "cucumber"]))
pths = list(map(f_partial, indices[0]))
```

```python
pths = [pths[i] for i in [1, 8,7,  0, 2, 3, 4, 5, 6]]
```

```python
f, axes = plt.subplots(
    1, 3, figsize=(3*4.5, 3*1.5)
)
for i, ax in enumerate(axes.flat):
    img = Image.open(pths[i])
    ax.imshow(img)
    _ = ax.axis("off")
```
